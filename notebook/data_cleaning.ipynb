### Cleaning log helper
Every fix below calls `log()`. The full log is displayed in Section 5.

```python
LOG = []

def log(table, issue, rows_affected, action, reason):
    LOG.append({'step': len(LOG)+1, 'table': table, 'issue': issue,
                'rows_affected': int(rows_affected), 'action': action, 'reason': reason})

log('customers', 'Initial row count', len(customers), 'none', 'baseline')
log('orders', 'Initial row count', len(orders), 'none', 'baseline')
```
